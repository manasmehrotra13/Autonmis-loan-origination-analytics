```python
import uuid
import random
import datetime
import pandas as pd
from faker import Faker
import os

faker = Faker()

# Function to generate random dates
def random_date(start, end):
    return start + datetime.timedelta(days=random.randint(0, (end - start).days))

# Generate mock data for each table
num_records = 2000
regions = [str(uuid.uuid4()) for _ in range(5)]
channels = [str(uuid.uuid4()) for _ in range(num_records)]
agents = [str(uuid.uuid4()) for _ in range(num_records)]
customers = [str(uuid.uuid4()) for _ in range(num_records)]
loans = [str(uuid.uuid4()) for _ in range(num_records)]
application_statuses = [str(uuid.uuid4()) for _ in range(num_records)]
decline_reasons = [str(uuid.uuid4()) for _ in range(num_records)]

# Fact Leads
df_fact_leads = pd.DataFrame({
    "lead_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "customer_id": random.choices(customers, k=num_records),
    "lead_source": random.choices(["Online", "Referral", "Walk-in", "Call Center", "Email", "Social Media", "Mobile App"], k=num_records),
    "created_date": [random_date(datetime.date(2025, 1, 1), datetime.date(2025, 12, 31)) for _ in range(num_records)],
    "lead_status": random.choices(["New", "Contacted", "Qualified", "Converted", "Pending", "Rejected", "Approved"], k=num_records),
    "conversion_flag": random.choices([True, False], k=num_records),
    "channel_id": random.choices(channels, k=num_records),
    "region_id": random.choices(regions, k=num_records)
})

# Fact Applications
df_fact_applications = pd.DataFrame({
    "application_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "customer_id": random.choices(customers, k=num_records),
    "loan_id": random.choices(loans, k=num_records),
    "applied_date": [random_date(datetime.date(2025, 1, 1), datetime.date(2025, 12, 31)) for _ in range(num_records)],
    "application_status_id": random.choices(application_statuses, k=num_records),
    "loan_amount": [round(random.uniform(5000, 50000), 2) for _ in range(num_records)],
    "tenure": random.choices(range(6, 60), k=num_records),
    "interest_rate": [round(random.uniform(5.0, 15.0), 2) for _ in range(num_records)],
    "region_id": random.choices(regions, k=num_records),
    "channel_id": random.choices(channels, k=num_records),
    "agent_id": random.choices(agents, k=num_records)
})

# Fact Underwriting
df_fact_underwriting = pd.DataFrame({
    "underwriting_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "application_id": random.choices(df_fact_applications["application_id"], k=num_records),
    "risk_score": [round(random.uniform(0, 1), 2) for _ in range(num_records)],
    "decision_date": [random_date(datetime.date(2025, 1, 1), datetime.date(2025, 12, 31)) for _ in range(num_records)],
    "approval_status": random.choices(["Approved", "Declined", "Pending", "In Review", "On Hold", "Escalated", "Withdrawn"], k=num_records),
    "requested_amount": [round(random.uniform(5000, 50000), 2) for _ in range(num_records)],
    "approved_amount": [round(random.uniform(5000, 50000), 2) for _ in range(num_records)],
    "credit_score": random.choices(range(250, 950), k=num_records),
    "processing_time_days": random.choices(range(1, 30), k=num_records)
})

# Fact Disbursements
df_fact_disbursements = pd.DataFrame({
    "disbursement_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "application_id": random.choices(df_fact_applications["application_id"], k=num_records),
    "disbursed_amount": [round(random.uniform(5000, 50000), 2) for _ in range(num_records)],
    "disbursement_date": [random_date(datetime.date(2025, 1, 1), datetime.date(2025, 12, 31)) for _ in range(num_records)],
    "loan_start_date": [random_date(datetime.date(2025, 1, 1), datetime.date(2025, 12, 31)) for _ in range(num_records)],
    "loan_end_date": [random_date(datetime.date(2026, 1, 1), datetime.date(2026, 12, 31)) for _ in range(num_records)],
    "repayment_term_months": random.choices(range(6, 60), k=num_records),
    "interest_rate": [round(random.uniform(5.0, 15.0), 2) for _ in range(num_records)]
})

# Fact Declines
df_fact_declines = pd.DataFrame({
    "decline_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "application_id": random.choices(df_fact_applications["application_id"], k=num_records),
    "customer_id": random.choices(customers, k=num_records),
    "decline_reason_id": random.choices(decline_reasons, k=num_records),
    "decline_date": [random_date(datetime.date(2025, 1, 1), datetime.date(2025, 12, 31)) for _ in range(num_records)],
    "credit_score_at_decline": random.choices(range(250, 950), k=num_records),
    "income_at_decline": [round(random.uniform(20000, 200000), 2) for _ in range(num_records)],
    "loan_amount_requested": [round(random.uniform(5000, 50000), 2) for _ in range(num_records)],
    "agent_id": random.choices(agents, k=num_records)
})

# Fact Channel Performance
df_fact_channel_performance = pd.DataFrame({
    "performance_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "channel_id": random.choices(channels, k=num_records),
    "leads_generated": random.choices(range(10, 500), k=num_records),
    "applications_submitted": random.choices(range(5, 300), k=num_records),
    "conversion_rate": [round(random.uniform(0.1, 1.0), 2) for _ in range(num_records)],
    "cost_per_lead": [round(random.uniform(10, 500), 2) for _ in range(num_records)],
    "revenue_generated": [round(random.uniform(1000, 100000), 2) for _ in range(num_records)]
})

# Fact Customer Feedback
df_fact_customer_feedback = pd.DataFrame({
    "feedback_id": [str(uuid.uuid4()) for _ in range(num_records)],
    "customer_id": random.choices(customers, k=num_records),
    "application_id": random.choices(df_fact_applications["application_id"], k=num_records),
    "agent_id": random.choices(agents, k=num_records),
    "feedback_score": random.choices([1, 2, 3, 4, 5], k=num_records),
    "processing_time_satisfaction": random.choices(["Very Satisfied", "Satisfied", "Neutral", "Dissatisfied", "Very Dissatisfied", "Extremely Satisfied", "Somewhat Dissatisfied"], k=num_records),
    "communication_satisfaction": random.choices(["Very Satisfied", "Satisfied", "Neutral", "Dissatisfied", "Very Dissatisfied", "Extremely Satisfied", "Somewhat Dissatisfied"], k=num_records),
    "ease_of_application": random.choices(["Very Easy", "Easy", "Neutral", "Difficult", "Very Difficult", "Somewhat Easy", "Somewhat Difficult"], k=num_records),
    "improvement_suggestions": random.choices(["Faster processing", "Better communication", "More loan options", "Lower interest rates", "Easier documentation", "Flexible repayment terms", "Improved customer support"], k=num_records)
})

# Dimension Customers
df_dim_customers = pd.DataFrame({
    "customer_id": customers,
    "first_name": [faker.first_name() for _ in range(num_records)],
    "last_name": [faker.last_name() for _ in range(num_records)],
    "dob": [faker.date_of_birth(minimum_age=18, maximum_age=65) for _ in range(num_records)],
    "gender": random.choices(["Male", "Female", "Other"], k=num_records),
    "income": [round(random.uniform(20000, 200000), 2) for _ in range(num_records)],
    "employment_status": random.choices(["Employed", "Self-Employed", "Unemployed", "Student", "Retired", "Part-Time", "Contract"], k=num_records),
    "credit_score": random.choices(range(250, 950), k=num_records),
    "address": [faker.address() for _ in range(num_records)],
    "region_id": random.choices(regions, k=num_records)
})

# Dimension Loans
df_dim_loans = pd.DataFrame({
    "loan_id": loans,
    "loan_type": random.choices(["Personal", "Home", "Auto", "Education", "Business", "Medical", "Debt Consolidation"], k=num_records),
    "interest_rate": [round(random.uniform(5.0, 15.0), 2) for _ in range(num_records)],
    "max_tenure": random.choices(range(12, 240), k=num_records),
    "min_amount": [round(random.uniform(1000, 5000), 2) for _ in range(num_records)],
    "max_amount": [round(random.uniform(50000, 500000), 2) for _ in range(num_records)],
    "eligibility_criteria": random.choices(["Minimum income 25K", "Credit Score 650+", "Stable employment required", "Age 21-60", "No past loan defaults", "Valid ID proof required", "Minimum 6 months in current job"], k=num_records)
})

# Dimension Application Status
df_dim_application_status = pd.DataFrame({
    "status_id": application_statuses,
    "status_name": random.choices(["Pending", "Approved", "Rejected", "Under Review", "Incomplete", "On Hold", "Needs More Info"],k=num_records),
    "description": random.choices(["Waiting for review", "Loan approved", "Loan rejected", "In review process", "Incomplete details"], k=num_records),
    "is_final_status": random.choices([False, True], k=num_records)
})

# Dimension Regions
df_dim_regions = pd.DataFrame({
    "region_id": regions,
    "country": random.choices(["India", "USA", "UK", "Canada", "Australia"], k=5),
    "state": [faker.state() for _ in range(5)],
    "city": [faker.city() for _ in range(5)],
    "postal_code": [faker.postcode() for _ in range(5)]
})

# Dimension Channels
df_dim_channels = pd.DataFrame({
    "channel_id": channels,
    "channel_name": random.choices(["Online", "Branch", "Call Center", "Agent Referral", "Mobile App", "Website Chat", "Social Media"], k=num_records),
    "channel_type": random.choices(["Digital", "Physical", "Agent", "Call Center", "Branch", "Mobile App", "Third-Party"], k=num_records),
    "is_digital": random.choices([True, False], k=num_records),
    "cost_per_acquisition": [round(random.uniform(10, 500), 2) for _ in range(num_records)]
})

# Dimension Decline Reasons
df_dim_decline_reasons = pd.DataFrame({
    "reason_id": decline_reasons,
    "reason_name": random.choices(
        ["Low Credit Score", "Insufficient Income", "High Debt-to-Income Ratio",
         "Unverified Employment", "Incomplete Application", "Negative Credit History",
         "Policy Restrictions"], k=num_records),
    "risk_category": random.choices(
        ["High", "Medium", "Low", "Very High", "Very Low", "Moderate", "Critical"], k=num_records),
    "mitigation_strategy": random.choices(
        ["Improve credit score", "Increase income", "Reduce debts",
         "Provide employment proof", "Complete application", "Offer collateral",
         "Seek co-applicant"], k=num_records)
})

# Dimension Agents
df_dim_agents = pd.DataFrame({
    "agent_id": agents,
    "first_name": [faker.first_name() for _ in range(num_records)],
    "last_name": [faker.last_name() for _ in range(num_records)],
    "region_id": random.choices(regions, k=num_records),
    "assigned_channel_id": random.choices(channels, k=num_records),
    "experience_years": random.choices(range(1, 20), k=num_records),
    "performance_rating": random.choices(["Excellent", "Very Good", "Good", "Above Average", "Average", "Poor", "Very Poor"], k=num_records)
})

# Generate and save all tables
all_tables = {
    "fact_leads": df_fact_leads,
    "fact_applications": df_fact_applications,
    "fact_underwriting": df_fact_underwriting,
    "fact_disbursements": df_fact_disbursements,
    "fact_declines": df_fact_declines,
    "fact_channel_performance": df_fact_channel_performance,
    "fact_customer_feedback": df_fact_customer_feedback,
    "dim_customers": df_dim_customers,
    "dim_loans": df_dim_loans,
    "dim_application_status": df_dim_application_status,
    "dim_regions": df_dim_regions,
    "dim_channels": df_dim_channels,
    "dim_decline_reasons": df_dim_decline_reasons,
    "dim_agents": df_dim_agents
}

data_folder = r"C:\Users\manas\OneDrive\Desktop\Data"

# Ensure the directory exists
os.makedirs(data_folder, exist_ok=True)

# Save as CSV files in the 'Data' folder
for table_name, df in all_tables.items():
    file_path = os.path.join(data_folder, f"{table_name}.csv")
    df.to_csv(file_path, index=False)

print("Mock data generated successfully!")
```
