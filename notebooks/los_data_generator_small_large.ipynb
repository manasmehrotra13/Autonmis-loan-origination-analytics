```python
import uuid
import random
import datetime
import pandas as pd
from faker import Faker
import os

faker = Faker()

# Function to generate random dates
def random_date(start, end):
    return start + datetime.timedelta(days=random.randint(0, (end - start).days))

# Generate mock data for each table
num_records_small = 1000
num_records_large=50000

regions = [str(uuid.uuid4()) for _ in range(5)]
channels = [str(uuid.uuid4()) for _ in range(num_records_large)]
agents = [str(uuid.uuid4()) for _ in range(num_records_large)]
customers = [str(uuid.uuid4()) for _ in range(num_records_large)]
loans = [str(uuid.uuid4()) for _ in range(num_records_large)]
application_statuses = [str(uuid.uuid4()) for _ in range(num_records_small)]
decline_reasons = [str(uuid.uuid4()) for _ in range(num_records_small)]

# Fact Leads
df_fact_leads = pd.DataFrame({
    "lead_id": [str(uuid.uuid4()) for _ in range(num_records_large)],
    "customer_id": random.choices(customers, k=num_records_large),
    "lead_source": random.choices(["Online", "Referral", "Walk-in", "Call Center", "Email", "Social Media", "Mobile App"], k=num_records_large),
    "created_date": [random_date(datetime.date(2024, 1, 1), datetime.date(2024, 12, 31)) for _ in range(num_records_large)],
    "lead_status": random.choices(["New", "Contacted", "Qualified", "Converted", "Pending", "Rejected", "Approved"], k=num_records_large),
    "conversion_flag": random.choices([True, False], k=num_records_large),
    "channel_id": random.choices(channels, k=num_records_large),
    "region_id": random.choices(regions, k=num_records_large)
})

# Fact Applications
df_fact_applications = pd.DataFrame({
    "application_id": [str(uuid.uuid4()) for _ in range(num_records_large)],
    "customer_id": random.choices(customers, k=num_records_large),
    "loan_id": random.choices(loans, k=num_records_large),
    "applied_date": [random_date(datetime.date(2024, 1, 1), datetime.date(2024, 12, 31)) for _ in range(num_records_large)],
    "application_status_id": random.choices(application_statuses, k=num_records_large),
    "loan_amount": [round(random.uniform(5000, 50000), 2) for _ in range(num_records_large)],
    "tenure": random.choices(range(6, 60), k=num_records_large),
    "interest_rate": [round(random.uniform(5.0, 15.0), 2) for _ in range(num_records_large)],
    "region_id": random.choices(regions, k=num_records_large),
    "channel_id": random.choices(channels, k=num_records_large),
    "agent_id": random.choices(agents, k=num_records_large)
})


# Dimension Customers
df_dim_customers = pd.DataFrame({
    "customer_id": customers,
    "first_name": [faker.first_name() for _ in range(num_records_large)],
    "last_name": [faker.last_name() for _ in range(num_records_large)],
    "dob": [faker.date_of_birth(minimum_age=18, maximum_age=65) for _ in range(num_records_large)],
    "gender": random.choices(["Male", "Female", "Other"], k=num_records_large),
    "income": [round(random.uniform(20000, 200000), 2) for _ in range(num_records_large)],
    "employment_status": random.choices(["Employed", "Self-Employed", "Unemployed", "Student", "Retired", "Part-Time", "Contract"], k=num_records_large),
    "credit_score": random.choices(range(250, 950), k=num_records_large),
    "address": [faker.address() for _ in range(num_records_large)],
    "region_id": random.choices(regions, k=num_records_large)
})

# Dimension Loans
df_dim_loans = pd.DataFrame({
    "loan_id": loans,
    "loan_type": random.choices(["Personal", "Home", "Auto", "Education", "Business", "Medical", "Debt Consolidation"], k=num_records_large),
    "interest_rate": [round(random.uniform(5.0, 15.0), 2) for _ in range(num_records_large)],
    "max_tenure": random.choices(range(12, 240), k=num_records_large),
    "min_amount": [round(random.uniform(1000, 5000), 2) for _ in range(num_records_large)],
    "max_amount": [round(random.uniform(50000, 500000), 2) for _ in range(num_records_large)],
    "eligibility_criteria": random.choices(["Minimum income 25K", "Credit Score 650+", "Stable employment required", "Age 21-60", "No past loan defaults", "Valid ID proof required", "Minimum 6 months in current job"], k=num_records_large)
})

# Dimension Decline Reasons
df_dim_decline_reasons = pd.DataFrame({
    "reason_id": decline_reasons,
    "reason_name": random.choices(
        ["Low Credit Score", "Insufficient Income", "High Debt-to-Income Ratio",
         "Unverified Employment", "Incomplete Application", "Negative Credit History",
         "Policy Restrictions"], k=num_records_small),
    "risk_category": random.choices(
        ["High", "Medium", "Low", "Very High", "Very Low", "Moderate", "Critical"], k=num_records_small),
    "mitigation_strategy": random.choices(
        ["Improve credit score", "Increase income", "Reduce debts",
         "Provide employment proof", "Complete application", "Offer collateral",
         "Seek co-applicant"], k=num_records_small)
})

# Generate and save all tables
all_tables = {
    "fact_leads": df_fact_leads,
    "fact_applications": df_fact_applications,
    "dim_customers": df_dim_customers,
    "dim_loans": df_dim_loans,
    "dim_decline_reasons": df_dim_decline_reasons,
}

data_folder = r"C:\Users\manas\OneDrive\Desktop\Tables"

# Ensure the directory exists
os.makedirs(data_folder, exist_ok=True)

# Save as CSV files in the 'Data' folder
for table_name, df in all_tables.items():
    file_path = os.path.join(data_folder, f"{table_name}.csv")
    df.to_csv(file_path, index=False)

print("Mock data generated successfully!")
```
